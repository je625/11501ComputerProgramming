## 二、題目說明（Problem Description）

請設計一個 Python 程式，建立學生分數分析系統。

程式必須能夠接收多筆分數，根據指定的倍率（multiplier）及加分值（bonus）調整分數，接著計算調整後的分數平方、總分及平均分數，最後回傳分析結果。

### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

### Part 2：呼叫函數

## 三、語法整理（Learning Focus）

| 語法 | 用途 |
|---|---|
| `*scores`（函數定義） | 將不定數量的位置參數收集成 tuple |
| `**options`（函數定義） | 將不定數量的關鍵字參數收集成 dictionary |
| `score * multiplier` | 乘法 |
| `score ** 2` | 次方 |
| `f(*scores)` | 將可迭代物件解包為位置參數 |
| `f(**settings)` | 將字典解包為關鍵字參數 |
| `return` | 將函數的結果回傳給呼叫端 |

## 四、總結
請根據本題，說明 `*`、`**` 與 `return` 在 Python 中的不同用途，並舉出至少一個程式碼例子，展示同一個符號在不同語法位置所代表的意義。

* 在函數定義中表示接收不定數量的位置參數。例如本題中的 *scores，會把所有傳入的分數收集成一個 tuple（元組）。但當 * 出現在函數呼叫時，則代表解包（unpacking）的意思，可以將串列或 tuple 中的元素逐一拆開後傳入函數。因此，同樣都是 *，在函數定義與函數呼叫中的用途並不相同。
def show_scores(*scores):
    print(scores)

show_scores(60, 70, 80)


** 的作用與 * 類似，但它是用來處理關鍵字參數。在函數定義中的 **options，會把所有關鍵字參數收集成一個 dict（字典）。而在函數呼叫時，** 則可以將字典中的鍵和值解包成關鍵字參數傳入函數。因此，同一個 ** 符號在不同位置也具有不同的意義。
def show_info(**info):
    print(info)

show_info(name="Tom", age=18)
至於 return，它的用途是將函數計算完成的結果回傳給函數外部使用。在本題中，函數最後利用 return 回傳一個包含調整後分數、平方分數、總分和平均分數的字典。呼叫函數的程式可以將這個回傳值存入變數，之後再繼續使用。如果函數只有使用 print()，雖然可以把結果顯示在螢幕上，但無法方便地讓其他程式碼繼續使用這些結果，因此在設計函數時通常會使用 return 來回傳資料。
def add(a, b):
return a + b
 
result = add(3, 5)
 
print(result)
